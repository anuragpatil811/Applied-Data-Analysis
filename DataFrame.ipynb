```python
import pandas as pd
```

```python
data = {
    "Name": ["Amit", "Priya", "Rahul", "Neha"],
    "Age": [21, 22, 30, 54],
    "Marks": [85, 90, 78, 92]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
print(df["Marks"])
```

```python
print(df[["Name", "Marks"]])
```

```python
high_scorer = df[df["Marks"]>80]
high_scorer
```

```python
df["Pass"] = df["Marks"] > 40
df
```
