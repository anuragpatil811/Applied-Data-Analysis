```python
import pandas as pd
```

```python
data = {
    "Region": ["North", "South", "East", "West", "North", "South"],
    "Product": ["A", "B", "A", "B", "A", "B"],
    "Sales": [15000, 18000, 12000, 25000, 17000, 22000]
}
```

```python
df = pd.DataFrame(data)
```

```python
pivot_table = df.pivot_table(values="Sales",index="Region", columns="Product", aggfunc="sum")
```

```python
pivot_table
```

```python
pivot_table = df.pivot_table(values="Sales", index="Region", aggfunc=["sum", "mean", "count"])
pivot_table
```
