```python
import numpy as np
salary_data = pd.DataFrame({"Salary": [50000, 55000, 60000, 1000000,
65000]})
```

```python
mean = np.mean(salary_data["Salary"])
std_dev = np.std(salary_data["Salary"])
```

```python
outlier_threshold = mean + 3*std_dev
```

```python
cleaned_salary = salary_data[salary_data["Salary"]<outlier_threshold]
cleaned_salary
```
